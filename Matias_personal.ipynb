```python
def import_txt_files_to_array(folder_path):
    """
    Import all .txt files from a folder into a list of strings,
    including only the text after the first occurrence of "\n\n\n".

    Args:
        folder_path (str): Path to the folder containing .txt files.

    Returns:
        list: A list where each element is the content of a .txt file (after "\n\n\n").
    """
    txt_files_content = []

    for filename in os.listdir(folder_path):
        if filename.endswith('.txt'):
            file_path = os.path.join(folder_path, filename)
            try:
                with open(file_path, 'r', encoding='utf-8') as file:
                    content = file.read()
                    # Split at the first occurrence of "\n\n\n" and take the part after it
                    parts = content.split('\n\n\n', 1)
                    if len(parts) > 1:
                        txt_files_content.append(parts[1])
                    else:
                        # If "\n\n\n" is not found, include the entire content
                        txt_files_content.append(content)
            except Exception as e:
                print(f"Error reading {filename}: {e}")

    return txt_files_content

###########################################################################################

folder_path = r'C:\Users\Geniet\AppliedML\AppML_Project\ML2026_Project\Speeches'  # Replace with your folder path
txt_array = import_txt_files_to_array(folder_path)

# Print the first 100 characters of each file to verify
for i, content in enumerate(txt_array):
    print(f"File {i + 1} (first 150 chars): {content[:150]}...")
```
