```python
import numpy as np
from scipy import sparse
from pathlib import Path

# --------------------------------------------------------
# LOAD ARTICLE DATA + LABEL FILES
# --------------------------------------------------------

base_dir = Path("Data_files")

data_all = []
labels_all = []

for i in range(19):

    idx = f"{i:02d}"

    data_file = base_dir / f"data_articles_{idx}.npz"
    label_file = base_dir / f"labels_articles_{idx}.txt"

    # skip if either file does not exist
    if not data_file.exists() or not label_file.exists():

        print(f"Skipping {idx} (missing file)")
        continue

    print(f"Loading {data_file.name} and {label_file.name}")

    X = sparse.load_npz(data_file)
    with open(label_file, "r", encoding="latin1") as f:
        y = np.array([line.strip() for line in f if line.strip()])

    print("  data shape :", X.shape)
    print("  labels shape:", y.shape)

    data_all.append(X)
    labels_all.append(y)

# --------------------------------------------------------
# CHECK SOMETHING WAS LOADED
# --------------------------------------------------------

if len(data_all) == 0:
    raise ValueError("No datasets were loaded.")

# --------------------------------------------------------
# CREATE COMBINED DATASET
# --------------------------------------------------------

data_articles_all = sparse.vstack(data_all, format="csr")
labels_articles_all = np.concatenate(labels_all)

# --------------------------------------------------------
# PRINT COMBINED SHAPES
# --------------------------------------------------------

print("\nCombined dataset:")
print(data_articles_all.shape)
print(labels_articles_all.shape)

# --------------------------------------------------------
# SAVE COMBINED DATASET
# --------------------------------------------------------

sparse.save_npz(
    base_dir / "data_articles_all.npz",
    data_articles_all
)

with open(base_dir / "labels_articles_all.txt", "w", encoding="utf-8") as f:
    for label in labels_articles_all:
        f.write(f"{label}\n")

print("\nAll article data saved!")
```
